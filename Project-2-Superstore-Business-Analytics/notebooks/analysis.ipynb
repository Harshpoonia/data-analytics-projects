```python
# Missing value

df.isnull().sum()
```

```python
# Duplicate Records

df.duplicated().sum()
```

```python
# Data types

df.dtypes
```

```python
# Convert Date Columns

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df[['Order Date', 'Ship Date']].dtypes

```

```python
# Check Date Range

print("Start Date:", df['Order Date'].min())
print("End Date:", df['Order Date'].max())
```

## 5. Feature Engineering

```python
# Create Year

df['Year'] = df['Order Date'].dt.year
```

```python
# Create Month

df['Month'] = df['Order Date'].dt.month_name()
```

```python
# Create Month Number

df['Month Number'] = df['Order Date'].dt.month
```

```python
# Create Quarter

df['Quarter'] = df['Order Date'].dt.quarter
```

```python
# Create Shipping Days

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days
```

```python
# Create Profit Margin

df['Profit Margin (%)'] = (
    df['Profit'] / df['Sales']
) * 100
```

```python
# New Features

df[['Year',
    'Month',
    'Month Number',
    'Quarter',
    'Shipping Days',
    'Profit Margin (%)']].head()
```

## 6. Exploratory Data Analysis

### 6.1 : Sales by Region 

### Objective

Analyze sales performance across regions to identify which geographical areas contribute the highest revenue and support business growth.

```python
# Calculate total sales by region
region_sales = (
    df.groupby('Region')['Sales']
    .sum()
    .sort_values(ascending=False)
)

# Create visualization
plt.figure(figsize=(8, 5))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

# Add chart title and labels
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')

# Improve layout
plt.tight_layout()

# Save chart for README and portfolio
plt.savefig(
    '../images/sales_by_region.png',
    dpi=300,
    bbox_inches='tight'
)

# Display chart
plt.show()

# Display sales values
print(region_sales)
```

### Insight

The West region generated the highest sales revenue of **$725,457.82**, making it the strongest-performing region in the business. The East region followed closely with **$678,781.24** in total sales.

In contrast, the South region recorded the lowest sales revenue of **$391,721.91**, indicating comparatively lower market performance.

The results reveal a noticeable variation in revenue contribution across regions, suggesting that business growth opportunities may exist in underperforming regions while successful strategies from the West region could be replicated elsewhere.

### 6.9 : Discount vs Profit

### Objective

Investigate the relationship between discounts and profit to determine how discounting strategies impact business profitability.(Question:"Are our discounts helping the business or hurting profitability?")

```python
# Create visualization
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit'
)

# Add chart title and labels
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

# Improve layout
plt.tight_layout()

# Save chart for README and portfolio
plt.savefig(
    '../images/discount_vs_profit.png',
    dpi=300,
    bbox_inches='tight'
)

# Display chart
plt.show()

# Calculate average profit by discount level
discount_profit = (
    df.groupby('Discount')['Profit']
    .mean()
    .sort_index()
)

print(discount_profit.round(2))
```

### Insight

The analysis reveals a strong negative relationship between discount levels and profitability.

Products sold with discounts between **0% and 20%** generally maintain positive average profits. However, profitability declines significantly as discount rates increase.

At a **30% discount**, average profit becomes negative (**-$45.68**), indicating that the business begins losing money on average transactions. The losses become substantially larger at higher discount levels, reaching **-$310.70** at a 50% discount.

These findings suggest that aggressive discounting strategies are negatively impacting profitability and may be eroding profit margins. While discounts can help drive sales volume, excessive discount levels appear to generate losses rather than sustainable revenue growth.

The business should carefully evaluate its discount policies and identify product categories or regions where high discounts are being applied without producing profitable outcomes.

### 6.10 : Correlation Heatmap

### Objective

Analyze the relationships between numerical variables to identify factors that influence sales, profit, quantity, discounts, and overall business performance.

```python
# Select numerical columns
numerical_cols = [
    'Sales',
    'Quantity',
    'Discount',
    'Profit',
    'Shipping Days',
    'Profit Margin (%)'
]

# Calculate correlation matrix
corr_matrix = df[numerical_cols].corr()

# Create visualization
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

# Add chart title
plt.title('Correlation Heatmap')

# Improve layout
plt.tight_layout()

# Save chart for README and portfolio
plt.savefig(
    '../images/correlation_heatmap.png',
    dpi=300,
    bbox_inches='tight'
)

# Display chart
plt.show()

# Display correlation matrix
print(corr_matrix.round(2))
```

### Insight

The correlation analysis highlights several important relationships between key business metrics.

Sales and Profit exhibit a moderate positive correlation (**0.48**), indicating that higher sales generally contribute to higher profits. However, the relationship is not perfectly linear, suggesting that factors such as discounts and product margins also influence profitability.

Discount and Profit show a negative correlation (**-0.22**), confirming earlier findings that higher discounts tend to reduce profitability.

The strongest relationship in the analysis is between Discount and Profit Margin (%) with a correlation of **-0.86**. This strong negative correlation indicates that increasing discounts significantly reduces profit margins and represents one of the most influential factors affecting business performance.

Shipping Days shows almost no correlation with sales or profit, suggesting that delivery duration has minimal direct impact on revenue and profitability within this dataset.

Overall, the analysis indicates that discount management plays a critical role in maintaining profitability and should be closely monitored when designing pricing and promotional strategies.

## 7. KPI Analysis

## Objective

Calculate key business performance indicators (KPIs) to evaluate overall sales performance, profitability, customer activity, and operational efficiency.

```python
# Calculate key business KPIs

total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order ID'].nunique()

total_customers = df['Customer ID'].nunique()

average_order_value = total_sales / total_orders

average_shipping_days = df['Shipping Days'].mean()

top_region = (
    df.groupby('Region')['Sales']
    .sum()
    .idxmax()
)

top_category = (
    df.groupby('Category')['Sales']
    .sum()
    .idxmax()
)

best_month = (
    df.groupby('Month')['Sales']
    .sum()
    .idxmax()
)

```

```python
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Profit',
        'Total Orders',
        'Total Customers',
        'Average Order Value',
        'Average Shipping Days',
        'Top Region',
        'Top Category',
        'Best Month'
    ],
    'Value': [
        f"${total_sales:,.2f}",
        f"${total_profit:,.2f}",
        total_orders,
        total_customers,
        f"${average_order_value:,.2f}",
        f"{average_shipping_days:.2f}",
        top_region,
        top_category,
        best_month
    ]
})

kpi_summary
```

## Insight

The business generated total sales of **$2,297,200.86** and total profit of **$286,397.02** across **5,009 unique orders** from **793 customers**.

The average order value was **$458.61**, indicating a relatively strong revenue contribution per transaction. Additionally, the average shipping time was **3.96 days**, suggesting an efficient order fulfillment process.

The **West** region emerged as the highest revenue-generating region, while **Technology** was identified as the top-performing product category.

**November** recorded the highest sales performance among all months, reinforcing earlier findings regarding strong year-end demand and seasonal sales patterns.

Overall, the KPI analysis indicates a healthy business with strong revenue generation, profitable operations, and clear opportunities for strategic growth through regional expansion, category optimization, and customer retention initiatives.
