# 4. Data Cleaning


## 4.1 Missing Value Analysis
### Objective

Analyze missing values and determine the appropriate treatment strategy.


```python
# Missing values count
df.isnull().sum()
```

## 4.2 Missing Value Percentage
### Objective

Measure the percentage of missing values to determine their impact on the analysis.

```python
# Calculate missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage[missing_percentage > 0]
```

### Missing Value Treatment Strategy

#### Why We Should Not Drop These Rows

If the rows containing missing income values are removed:

* Total records would decrease from **2,240** to **2,216**.
* Valuable customer information would be lost.
* The dataset size would be unnecessarily reduced.

#### Key Considerations

* Only **24 records** contain missing income values.
* Missing values represent approximately **1.07%** of the dataset.
* Income is an important feature for customer and spending analysis.
* All remaining columns contain valid information for these customers.

#### Selected Approach

Instead of removing customer records, the missing income values will be replaced using **Median Imputation**.

Median imputation is preferred because:

* It preserves all customer records.
* It minimizes information loss.
* It is less sensitive to extreme income values than mean imputation.
* It maintains dataset quality for subsequent analysis.

#### Decision

**Missing income values will be replaced with the median income of the dataset rather than removing customer records.**


## 4.3 Handle Missing Income Values
### Objective

Handle missing income values to preserve customer records and maintain dataset completeness.

```python
# Fill missing income values with median income
df['Income'] = df['Income'].fillna(df['Income'].median())

# Verify missing values
df['Income'].isnull().sum()
```

### Insight

The Income column contained a small number of missing values. Since the missing percentage was low and income is an important customer attribute, missing values were replaced using the median income. This approach preserves customer records while minimizing the influence of extreme values.

## 4.4 Duplicate Records
### Objective

Verify the presence of duplicate records and assess their impact on the analysis.

```python
# Count duplicate records
df.duplicated().sum()
```

### Insight

No duplicate records were identified in the dataset. Therefore, no duplicate removal was required.

## Data Cleaning Summary

### Key Findings

# 5. Feature Engineering
## What is Feature Engineering?

Feature Engineering is the process of creating new variables from existing data to improve analysis and uncover deeper business insights.

These engineered features help transform raw customer and campaign data into meaningful metrics that can be used for customer profiling, spending analysis, and campaign performance evaluation.


## 5.1 Customer Age
### Objective

Create a customer age feature from the year of birth to support demographic analysis.

```python
# Create customer age feature
df['Age'] = 2026 - df['Year_Birth']

df[['Year_Birth', 'Age']].head()
```

### Insight

The Age feature was created to enable age-based customer analysis and identify demographic patterns in purchasing and campaign response behavior.

## 5.2 Total Spending
### Objective

Calculate total spending across all product categories to measure overall customer value.

```python
# Create total spending feature
df['Total_Spending'] = (
    df['MntWines']
    + df['MntFruits']
    + df['MntMeatProducts']
    + df['MntFishProducts']
    + df['MntSweetProducts']
    + df['MntGoldProds']
)

df[['Total_Spending']].head()
```

### Insight

The Total_Spending feature provides a consolidated measure of customer expenditure across all product categories and serves as a key indicator of customer value.

## 5.3 Total Purchases
### Objective

Calculate the total number of purchases made across all sales channels.

```python
# Create total purchases feature
df['Total_Purchases'] = (
    df['NumWebPurchases']
    + df['NumCatalogPurchases']
    + df['NumStorePurchases']
)

df[['Total_Purchases']].head()
```

### Insight

The Total_Purchases feature measures overall customer purchasing activity across web, catalog, and store channels.

## 5.4 Accepted Campaigns
### Objective

Calculate the number of marketing campaigns accepted by each customer.

```python
# Create campaign acceptance feature
df['Accepted_Campaigns'] = (
    df['AcceptedCmp1']
    + df['AcceptedCmp2']
    + df['AcceptedCmp3']
    + df['AcceptedCmp4']
    + df['AcceptedCmp5']
)

df[['Accepted_Campaigns']].head()
```

### Insight

The Accepted_Campaigns feature measures customer responsiveness to previous marketing campaigns and helps identify highly engaged customers.

## 5.5 Total Children
### Objective

Calculate the total number of children in the household.

```python
# Create total children feature
df['Total_Children'] = (
    df['Kidhome']
    + df['Teenhome']
)

df[['Total_Children']].head()
```

### Insight

The Total_Children feature helps analyze how household composition influences purchasing behavior and campaign response.

## 5.6 Customer Tenure
### Objective

Calculate customer tenure to measure the length of the customer relationship.

```python
# Convert customer enrollment date
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
```

```python
# Create customer tenure feature
df['Customer_Tenure'] = (
    pd.Timestamp('2026-01-01') - df['Dt_Customer']
).dt.days

df[['Customer_Tenure']].head()
```

### Insight

Customer tenure measures how long customers have been associated with the business and can be used to evaluate loyalty and engagement.

## Feature Engineering Summary

### Features Created

- Age
- Total_Spending
- Total_Purchases
- Accepted_Campaigns
- Total_Children
- Customer_Tenure

These engineered features provide a richer representation of customer demographics, purchasing behavior, and campaign engagement, enabling more meaningful analysis during the EDA phase.
