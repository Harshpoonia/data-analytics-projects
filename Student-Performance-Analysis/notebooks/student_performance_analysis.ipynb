## Exploratory Data Analysis


### 5.1 Creating Average Score Feature


### Objective

The dataset contains separate scores for Mathematics, Reading, and Writing. To evaluate the overall academic performance of each student, a new feature called **average_score** is created by calculating the mean of these three subject scores.

```python

df["average_score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
) / 3

df.head()
```

### Insight

A new feature named **average_score** was successfully created. This feature combines Mathematics, Reading, and Writing scores into a single metric, making it easier to compare overall student performance and perform further analysis.

### 5.2 Overall Student Performance


### Objective

This analysis aims to understand the overall academic performance of students by examining the average score, highest score, lowest score, and subject-wise average scores.

```python
# Overall Average Score

overall_avg = df["average_score"].mean()
overall_avg
```

```python
# Highest Average Score

highest_avg = df["average_score"].max()
```

```python
# Lowest Average Score

lowest_avg = df["average_score"].min()
```

```python
# Subject Wise Average Score

subject_avg = df[
    ["math score", "reading score", "writing score"]
].mean()
```

```python
#OUTPUT
print(f"Overall Average Score: {overall_avg:.2f}")
print(f"Highest Average Score: {highest_avg:.2f}")
print(f"Lowest Average Score: {lowest_avg:.2f}")
print("\nSubject-wise Average Scores:")
print(subject_avg)
```

### Insight

The overall average score of students is **67.77**, indicating a moderate level of academic performance across the dataset.

The highest average score achieved by a student is **100.00**, demonstrating exceptional performance across all three subjects, while the lowest average score is **9.00**, highlighting a significant variation in student achievement.

Among the three subjects, **Reading** has the highest average score (**69.17**), followed by **Writing** (**68.05**) and **Mathematics** (**66.09**). This suggests that students generally perform better in language-based subjects than in Mathematics.

### 5.3 Top Performing Students


### Objective

This analysis identifies the highest-performing students based on their overall average score. Examining top performers helps understand the level of academic excellence present in the dataset.

```python


df.sort_values(
    by="average_score",
    ascending=False
).head(10)
```

### Insight

The analysis of the top-performing students reveals that several students achieved exceptionally high average scores across Mathematics, Reading, and Writing. The highest-performing student attained an average score of 100, demonstrating perfect performance in all subjects.

The presence of multiple high-scoring students indicates that while the overall average performance is moderate, a subset of students consistently excels across all academic areas.

### 5.4 Gender-wise Performance


### Objective

This analysis examines the academic performance of students based on gender. The goal is to identify any differences in Mathematics, Reading, Writing, and overall average scores.

```python
gender_performance = df.groupby("gender")[
    ["math score", "reading score", "writing score", "average_score"]
].mean()

gender_performance
```

### Insight

The gender-wise performance analysis reveals noticeable differences across subjects.

Female students achieved a higher overall average score (**69.57**) compared to male students (**65.84**). Females also outperformed males in both Reading (**72.61 vs 65.47**) and Writing (**72.47 vs 63.31**) scores.

However, male students scored higher in Mathematics, with an average score of **68.73** compared to **63.63** for female students.

These results suggest that female students generally perform better in language-related subjects, while male students show stronger performance in Mathematics. Despite the higher Mathematics scores among males, female students demonstrate superior overall academic performance due to significantly higher Reading and Writing scores.

### 5.5 Test Preparation Impact

### Objective

This analysis investigates the impact of completing a test preparation course on student academic performance. The goal is to determine whether participation in the course is associated with higher scores across subjects.

```python
test_prep = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score", "average_score"]
].mean()

test_prep.round(2)
```

### Insight

The analysis shows a strong positive impact of test preparation on student performance.

Students who completed the test preparation course achieved an average score of **72.67**, whereas students who did not complete the course achieved an average score of **65.04**. This represents an improvement of approximately **7.63 points**.

Students who completed the course also scored higher across all subjects:
- Mathematics: 69.70 vs 64.08
- Reading: 73.89 vs 66.53
- Writing: 74.42 vs 64.50

These results suggest that participation in test preparation programs is associated with significantly better academic performance and may help students improve their understanding and examination skills.

### 5.6 Parental Education Impact

### Objective

This analysis examines the relationship between parental education level and student academic performance. The goal is to determine whether higher parental education is associated with better student outcomes.

```python
parent_edu = df.groupby("parental level of education")[
    ["average_score"]
].mean().sort_values(
    by="average_score",
    ascending=False
)

parent_edu.round(2)
```

### Insight

The analysis indicates a positive relationship between parental education level and student academic performance.

Students whose parents hold a **master's degree** achieved the highest average score (**73.60**), followed by those whose parents hold a **bachelor's degree** (**71.92**). In contrast, students whose parents have only a **high school education** recorded the lowest average score (**63.10**).

A gradual decline in student performance can be observed as parental education levels decrease. This trend suggests that higher parental education may contribute to a more supportive learning environment, greater academic guidance, and increased educational awareness.

Overall, the results indicate that parental education level is an important factor associated with student academic achievement.

### 5.7 Subject-wise Analysis

### Objective

This analysis compares student performance across Mathematics, Reading, and Writing to identify the strongest and weakest academic areas.

```python
subject_avg = df[
    ["math score", "reading score", "writing score"]
].mean()

subject_avg.round(2)
```

### Insight

The subject-wise analysis shows that students performed best in Reading, with an average score of 69.17. Writing was the second-highest scoring subject with an average score of 68.05, while Mathematics had the lowest average score of 66.09.

The results indicate that students generally demonstrate stronger performance in language-related subjects compared to Mathematics. This may suggest that additional academic support and practice could be beneficial in improving mathematical performance.

### 5.8 Correlation Analysis

### Objective

This analysis examines the relationship between scores in Mathematics, Reading, and Writing to determine whether strong performance in one subject is associated with strong performance in others.

```python
correlation_matrix = df[
    ["math score", "reading score", "writing score"]
].corr()

correlation_matrix
```

### Insight

The correlation analysis reveals a strong positive relationship between all three subjects.

Reading and Writing scores have the highest correlation (**0.95**), indicating that students who perform well in Reading tend to perform exceptionally well in Writing. This is expected because both subjects require similar language comprehension and communication skills.

Mathematics also shows a strong positive correlation with Reading (**0.82**) and Writing (**0.80**). These values suggest that students who perform well academically in one subject are generally likely to perform well in the others.

Overall, the results indicate that academic performance across subjects is closely related, with the strongest association observed between Reading and Writing.
